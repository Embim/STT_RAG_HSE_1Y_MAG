# Инстуркция по работе с векторной базой данных 

```python
import weaviate

client = weaviate.connect_to_local(host="localhost", port=8080)

# Список всех коллекций
collections = client.collections.list_all()
print("Коллекции:", list(collections.keys()))

# Просмотр объектов конкретной коллекции
collection = client.collections.get("VideoChunk")

# Количество объектов
result = collection.aggregate.over_all(total_count=True)
print("Всего объектов:", result.total_count)

# Первые 10 объектов
for item in collection.iterator():
    print(item.uuid, item.properties)

client.close()
```
